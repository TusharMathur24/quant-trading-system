## Purpose of Outlier Analysis

Outlier analysis is performed to:

- identify extreme price movements  
- assess distribution skewness  
- understand tail risk exposure  
- evaluate robustness of the trading strategy  

This step provides qualitative insight rather than model optimization.


```python
import matplotlib.pyplot as plt
import os
os.makedirs('plots', exist_ok=True)

if 'returns' in df.columns:
    plt.figure(figsize=(10, 6))
    plt.boxplot(df['returns'].dropna())
    plt.title('Returns Outlier Analysis')
    plt.ylabel('Returns')
    plt.savefig('plots/outlier_analysis.png')
    plt.close()
```
