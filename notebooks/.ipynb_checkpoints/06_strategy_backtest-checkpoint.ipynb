## Loading Feature Dataset

```python
df = load_csv('data/processed/nifty_features_5min.csv')
```

```python
df = df.dropna()
```

## Feature Selection for Strategy

A subset of features is selected to drive signal generation
and regime-conditioned trading logic.


```python
feats = ['avg_iv', 'iv_spread', 'ema_5']
valid_feats = [f for f in feats if f in df.columns]
```

```python
if not valid_feats: valid_feats = ['returns']
```

## Strategy Design

The trading strategy follows a regime-aware framework:

- market regimes are identified using the trained regime model  
- signals are generated conditionally based on regime state  
- positions are scaled according to volatility  
- risk management rules are applied at each step  


```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from hmmlearn.hmm import GaussianHMM

# 1. Prepare feature matrix
X = df[valid_feats].copy()
X = X.replace([np.inf, -np.inf], np.nan)
X = X.dropna()

# 2. Minimal data check
if X.shape[0] < 20:
    raise RuntimeError("Too few samples for regime detection")

# 3. Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Create and FIT HMM on the SAME data
model = GaussianHMM(
    n_components=3,
    covariance_type="diag",   # critical for stability
    n_iter=200,
    random_state=42
)

model.fit(X_scaled)

# 5. Predict regimes
regimes = model.predict(X_scaled)

# 6. Align back to dataframe
df_regime = df.loc[X.index].copy()
df_regime["regime"] = regimes

# 7. Validation output
print("Regime counts:")
print(df_regime["regime"].value_counts())
```

```python
df['regime'] = regimes
```

## Regime Mapping

The raw regime labels produced by the regime detection model are mapped
to simplified directional regimes to enable consistent trading logic.

The mapping is defined as:

- 0 → Neutral regime  
- 1 → Bullish regime  
- 2 → Bearish regime  

This transformation allows the strategy to apply directional rules
without relying on model-specific label ordering.


```python
mapping = {0:0, 1:1, 2:-1} # Simplified default
df['regime'] = df['regime'].map(mapping)
```

## Signal Generation

Trading signals are generated using the regime-aware strategy logic.

The signal generation module combines:

- detected market regime  
- feature-based conditions  
- predefined trading rules  

The resulting signal represents the trading position taken at each
time step.


```python
res = generate_signals(df)
```

## Performance Metrics

The strategy is evaluated using standard quantitative performance

Computed metrics include:

These metrics provide insight into both profitability and risk

```python
metrics = calculate_metrics(res)
print(metrics)
```

```python
import matplotlib.pyplot as plt
import os

os.makedirs("plots", exist_ok=True)

res['returns'] = res['returns'].fillna(0)
res['signal'] = res['signal'].fillna(0)

res['strategy_returns'] = res['signal'].shift(1) * res['returns']
res['strategy_returns'] = res['strategy_returns'].fillna(0)

res['cumulative_returns'] = (1 + res['strategy_returns']).cumprod()

plt.figure(figsize=(10, 6))
plt.plot(res['cumulative_returns'], label="Equity Curve")

plt.title("Strategy Equity Curve")
plt.xlabel("Time")
plt.ylabel("Cumulative Returns")

plt.legend()
plt.grid(True)

plt.savefig("plots/equity_curve.png")
plt.close()

print("Equity curve saved to plots/equity_curve.png")
```
