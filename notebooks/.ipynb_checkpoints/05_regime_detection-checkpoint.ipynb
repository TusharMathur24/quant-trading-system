## Module Imports

Regime detection logic is implemented using a Hidden Markov Model
to identify latent market states from observable features.

```python
from src.data_io import load_csv
from src.regime import RegimeModel
import os
```

## Loading Feature Dataset

The feature-enriched dataset generated in the previous step is used
as input for regime identification.


```python
df = load_csv('data/processed/nifty_features_5min.csv')
df = df.dropna()
```

## Feature Selection

A subset of features is selected to capture market structure and volatility.

These features are used to infer hidden regimes.


```python
feats = ['avg_iv', 'iv_spread', 'ema_5']
```

## Feature Availability Check

Selected features are validated to ensure availability before
model training. Missing features are handled gracefully.


```python
valid_feats = [f for f in feats if f in df.columns]
if not valid_feats:
    print('Warning: No features for regime. Using returns.')
    valid_feats = ['returns']
```

```python
X = df[valid_feats].values
```

## Regime Model Training

A Hidden Markov Model is trained on the selected feature set
to identify latent market regimes.

```python
model = RegimeModel()
```

```python
model.fit(X[:int(0.7*len(X))])
regimes = model.predict(X)
import pandas as pd; pd.Series(regimes).value_counts()
```

## Saving Regime Model

The trained regime model is saved for reuse during strategy backtesting.


```python
os.makedirs('../models', exist_ok=True)
```

```python
model.save('../models/regime_hmm.joblib')
```

```python
print('Saved regime model')
```

```python
# Plot Regime Distribution
import matplotlib.pyplot as plt

regimes = model.predict(X)
plt.figure(figsize=(10, 6))
plt.hist(regimes, bins=3, alpha=0.7, rwidth=0.8)
plt.title('Regime Distribution')
plt.xlabel('Regime State')
plt.ylabel('Frequency')
plt.xticks([0, 1, 2])
plt.grid(axis='y', alpha=0.5)
plt.savefig(f"{PLOTS_DIR}/regime_distribution.png")
plt.show()
plt.close()
```
