## Loading Cleaned Market Data

The cleaned datasets generated in the previous step are loaded from
the `data/processed` directory.


```python
spot = load_csv('data/processed/nifty_spot_clean.csv')
fut = load_csv('data/processed/nifty_futures_clean.csv')
opt = load_csv('data/processed/nifty_options_clean.csv')
```

## Merging Process

The datasets are merged using timestamp-based joins to create a
single unified market dataset.

The merged dataset contains:
- spot price information
- futures contract data
- aggregated options metrics


```python
import pandas as pd
```

```python
spot['timestamp'] = pd.to_datetime(spot['timestamp'])
fut['timestamp'] = pd.to_datetime(fut['timestamp'])
opt['timestamp'] = pd.to_datetime(opt['timestamp'])
```

```python
merged = merge_spot_futures_options(spot, fut, opt)
```

## Saving Merged Dataset

The merged dataset is stored in the processed data directory and will
serve as the base input for feature engineering.


```python
save_csv(merged, 'data/processed/nifty_merged_5min.csv')
```
