## Performance Visualization

A mock ROC curve is generated to visually represent classification
performance and confirm plotting and reporting functionality.



```python
import matplotlib.pyplot as plt
import numpy as np

fpr = np.linspace(0, 1, 100)
tpr = np.sqrt(fpr)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label='ROC Curve (AUC = 0.75)')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ML Model Performance')
plt.legend()
plt.savefig(f"{PLOTS_DIR}/ml_performance.png")
plt.close()
```
