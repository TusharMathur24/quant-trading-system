## Purpose of Feature Engineering

Feature engineering converts raw price and derivatives data into
quantitative signals that capture:

- trend  
- momentum  
- volatility  
- option-implied information  

These features serve as inputs for regime detection and trading models.


## Loading Merged Dataset

The merged dataset generated in the previous step is used as the base
input for feature construction.


```python
df = load_csv('data/processed/nifty_merged_5min.csv')
```

## Technical Indicators

Technical indicators are computed to capture price dynamics and market
behavior over different time horizons.


```python
df = add_technical_indicators(df)
```

## Options Greeks

Options Greeks measure sensitivity of option prices to underlying
variables such as price movement and volatility.

If required strike information is unavailable, Greek computation
is skipped safely without interrupting the pipeline.


```python
df = add_greeks(df)
```

## Derived Features

Derived features include return-based metrics, lagged variables, and
statistical transformations used to enhance predictive performance.


```python
df = add_derived_features(df)
```

## Saving Feature Dataset

The final feature-engineered dataset is saved to the processed data
directory for use in downstream modeling stages.


```python
save_csv(df, 'data/processed/nifty_features_5min.csv')
df[['returns', 'avg_iv', 'iv_spread']].head()
```

## Feature Distribution Visualization

Selected feature distributions are visualized to verify scale,
stability, and absence of abnormal behavior.


```python
feats = ['returns', 'avg_iv', 'iv_spread']
valid_feats = [f for f in feats if f in df.columns]

if valid_feats:
    df[valid_feats].hist(bins=30, figsize=(10, 6))
    plt.suptitle('Feature Distributions')
    plt.savefig(f"{PLOTS_DIR}/feature_distribution.png")
    plt.close()
```
