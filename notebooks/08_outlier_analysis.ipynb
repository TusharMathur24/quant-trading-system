## Purpose of Outlier Analysis

Outlier analysis is essential in quantitative finance to:

- detect abnormal price movements  
- identify extreme strategy returns  
- validate robustness of performance metrics  
- assess tail risk exposure  

This step ensures that reported metrics are not driven by rare anomalies.

## Returns Outlier Analysis

The distribution of strategy returns is analyzed to identify:

- unusually large gains  
- extreme losses  
- potential data errors  
- tail-risk behavior  


```python
import matplotlib.pyplot as plt

if 'returns' in df.columns:
    plt.figure(figsize=(10, 6))
    plt.boxplot(df['returns'].dropna())
    plt.title('Returns Outlier Analysis')
    plt.ylabel('Returns')
    plt.savefig(f"{PLOTS_DIR}/outlier_analysis.png")
    plt.close()
```
